`개념13` -- `range` 와 `append` 로 리스트 만들기

`-` `range(시작, 끝)` 은 시작부터 끝 바로 앞까지의 수를 차례로 준다.

````python
for n in range(3, 8):
    print(n)
````

`-` 빈 리스트(또는 처음 몇 개만 든 리스트)에 `append` 로 하나씩 붙여 나갈 수 있다. `x[-1]` 과 `x[-2]` 는 맨 끝, 끝에서 두 번째 원소다.

````python
an = [1, 1]
for n in range(3, 11):
    an.append(an[-1] + an[-2])
an
````
